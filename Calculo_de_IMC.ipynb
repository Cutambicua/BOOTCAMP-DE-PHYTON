```
# CÁLCULO DE IMC

peso = int(input('Digite o seu peso em Kg: '))
altura = float(input('Digite a sua altura em m/cm: '))

# Validação dos dados
if peso <= 0 or altura <= 0:
    print("Erro: Peso e altura devem ser maiores que zero.")
else:
    IMC = peso / (altura ** 2)

    if IMC < 18.5:
        print(f"IMC = {IMC:.2f} ➝ Abaixo do peso.")
    elif 18.5 <= IMC <= 24.9:
        print(f"IMC = {IMC:.2f} ➝ Peso Normal.")
    elif 25 <= IMC <= 29.9:
        print(f"IMC = {IMC:.2f} ➝ Sobrepeso.")
    else:
        print(f"IMC = {IMC:.2f} ➝ Obesidade.")
```
