```python
import random  # Importa o módulo random para gerar números aleatórios

# Solicita ao jogador o número de linhas do tabuleiro
n = int(input("Digite o número de linhas do tabuleiro: "))

# Solicita ao jogador o número de colunas do tabuleiro
m = int(input("Digite o número de colunas do tabuleiro: "))

# Cria o tabuleiro vazio (uma lista de listas preenchida com ".")
tabuleiro = [["." for _ in range(m)] for _ in range(n)]

# Solicita ao jogador onde deseja esconder o tesouro
linha_tesouro = int(input(f"Digite a linha do tesouro (0 a {n-1}): "))
coluna_tesouro = int(input(f"Digite a coluna do tesouro (0 a {m-1}): "))

# Verifica se a posição fornecida está dentro do tabuleiro
while linha_tesouro < 0 or linha_tesouro >= n or coluna_tesouro < 0 or coluna_tesouro >= m:
    print("Posição inválida! Tente novamente.")
    linha_tesouro = int(input(f"Digite a linha do tesouro (0 a {n-1}): "))
    coluna_tesouro = int(input(f"Digite a coluna do tesouro (0 a {m-1}): "))

# Coloca o tesouro no tabuleiro
tabuleiro[linha_tesouro][coluna_tesouro] = "T"

# Coloca uma armadilha em uma posição aleatória diferente do tesouro
while True:
    linha_armadilha = random.randint(0, n-1)  # Linha aleatória
    coluna_armadilha = random.randint(0, m-1)  # Coluna aleatória
    # Garante que a armadilha não fique na mesma posição do tesouro
    if linha_armadilha != linha_tesouro or coluna_armadilha != coluna_tesouro:
        tabuleiro[linha_armadilha][coluna_armadilha] = "X"
        break

# Exibe o tabuleiro na tela
print("\nTabuleiro:")
for linha in tabuleiro:   # Percorre cada linha
    print(" ".join(linha))  # Imprime cada célula separada por espaço

# Pergunta ao jogador onde ele acha que o tesouro está
palpite_linha = int(input(f"\nDigite sua tentativa de linha (0 a {n-1}): "))
palpite_coluna = int(input(f"Digite sua tentativa de coluna (0 a {m-1}): "))

# Verifica se o jogador acertou o tesouro
if palpite_linha == linha_tesouro and palpite_coluna == coluna_tesouro:
    print("🎉 Parabéns! Você encontrou o tesouro!")
# Verifica se caiu na armadilha
elif palpite_linha == linha_armadilha and palpite_coluna == coluna_armadilha:
    print("💀 Você caiu na armadilha!")
# Caso contrário, errou
else:
    print("❌ Não encontrou nada, tente novamente!")
```
