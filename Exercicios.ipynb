```
def calcular(a,b):
  produto = a*b
  if produto <= 1000:
    return produto
  else:
    return a+b

calcular(50,200)
```

```
numero = 75869
contador = 0

while numero > 0:
    numero //= 10   # elimina o último dígito
    contador += 1

print("Total de dígitos:", contador)
```
